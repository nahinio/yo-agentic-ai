## Model Integration with OpenAi, Google Gemini, GROQ

### OpenAi Model Integration

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")    # type: ignore

model = init_chat_model(
    model="gpt-5"
)

model
```

```python
# invoke the model with a query
response = model.invoke("Hello, can you tell me a joke?")
response
```

```python
response.content
```

Another way to call the model: ChatOpenAI

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-5"
)
response = model.invoke("Hello, can you tell me a dad joke?")
response.content
```

### Google Gemini Model Integration

```python
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")    # type: ignore

gmodel = init_chat_model(
    model="google_genai:gemini-3.5-flash"
)

response = gmodel.invoke("Hello, can you tell me a joke?")
response.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash"
)
response = model.invoke("Hello, can you tell me a dad joke?")
print(response.content[0]['text'])    #type: ignore
```

### GROQ Model Integration

```python
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")    # type: ignore

groq_model = init_chat_model(
    model="qwen/qwen3.8-27b",
    model_provider="groq"
)

groq_model
```

```python
response = groq_model.invoke("Hello, can you tell me a sad quote?")
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(
    model="qwen/qwen3.8-27b",
)
response = model.invoke("Hello, can you tell me a dad joke?")
response
```
