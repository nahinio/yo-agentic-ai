# Streaming And Batch

### Streaming
Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. We can use a loop to process each chunk in real-time.

```python
import os
from dotenv import load_dotenv
load_dotenv()

from langchain.chat_models import init_chat_model

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")    # type: ignore

model = init_chat_model(
    model="gpt-5"
)
```

```python

model.invoke("Write a paragraph about Lionel Messi's career highlights in 200 words.").content
```

```python
for chunk in model.stream("Write about Lionel Messi's career highlights"):
    print(chunk.text, end="", flush=True)
```

### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can
be done in parallel:

```python
response = model.batch([
    "Tell me about Batman.",
    "Tell me about PyGame",
    "What is worse? To live as a monster, or to die as a good man?"],
    config = {
        "max_concurrency": 5    # maximum number of concurrent requests to the model
    }
)

for response in response:
    print(response.content)
```
